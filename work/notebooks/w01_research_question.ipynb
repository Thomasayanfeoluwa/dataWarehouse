## 1. My lane (or freestyle) and why

**Lane:** Content Refresh & Declining Page Prioritization.

**Why:** Search traffic decays unpredictably across content portfolios. Rather than indiscriminately updating old content, this project builds a decision-support system to identify and prioritize declining pages so editorial resources are spent where traffic recovery potential is highest.



```python
import pandas as pd

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")
print(f"Total pages loaded: {len(df):,}")
```

## 2. The question: decision, action, cost of a wrong call

* **Decision:** Deciding which declining pages to rewrite or refresh first.
* **Action:** An SEO editor allocates engineering and writing hours to restructure and update targeted pages.
* **Cost of a wrong call:**
  * *False Positive:* Wasting editorial budget rewriting content that was already performing well.
  * *False Negative:* Ignoring high-traffic pages that are actively losing ranking authority.

  

```python
# Verify target label column and unit granularity
print("Target label distribution:")
print(df["trend_direction"].value_counts(normalize=True).round(3))
```

## 3. Quick look at the data (2-3 real numbers)

Based on baseline exploratory analysis of `data/raw/content_refresh_anonymized.csv`:
1. Declining (`down`) pages have a significantly lower median content age (~216 days) compared to growing (`up`, ~292 days) or stable pages (`stable`, ~300 days).
2. This reveals a counter-intuitive decay signal: newly published assets drop faster than established pages, making manual age-based heuristics unreliable.



```python
# Calculate median content age by trend direction
age_by_trend = df.groupby("trend_direction")["content_age_days"].median()
print("Median Content Age (Days) by Trend Direction:")
print(age_by_trend.round(0).to_string())
```

* **What this work CANNOT claim:** Does not claim causal proof, rank guarantees, or deterministic predictions of search engine algorithm changes.
